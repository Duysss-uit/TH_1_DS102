```python
import numpy as np 

def input_function():
    user_input = input("Enter a function in terms of x (e.g., 3*x[0]**2 + 2*x[1] + 1): ")
    f = eval("lambda x: " + user_input)
    return f
def input_vector():
    user_input = input("Enter a vector of initial guesses (e.g., [1.0, 2.0, 3.0]): ")
    x0 = np.array(eval(user_input))
    return x0

def gradient(x, f):
    h = 1e-5
    grad = np.zeros_like(x)
    for i in range(len(x)):
        x_plus_h = np.copy(x)
        x_minus_h = np.copy(x)
        x_plus_h[i] += h
        x_minus_h[i] -= h
        grad[i] = (f(x_plus_h) - f(x_minus_h)) / (2 * h)
    return grad    
def hessian(x, f):
    n = len(x)
    h = 1e-5
    H = np.zeros((n, n), dtype=float)
    for j in range(n):
        x_plus = x.copy()
        x_minus = x.copy()
        x_plus[j] += h
        x_minus[j] -= h
        grad_plus = gradient(x_plus, f)
        grad_minus = gradient(x_minus, f)
        H[:, j] = (grad_plus - grad_minus) / (2 * h)
    return 0.5* (H + H.T) 
def newton_raphson(f, x0, tol=1e-6, max_iter=100):
    x = x0
    for i in range(max_iter):
        grad = gradient(x, f)
        H = hessian(x, f)
        try:
            delta_x = np.linalg.solve(H, -grad)
        except np.linalg.LinAlgError:
            print("Hessian is singular. Stopping iteration.")
            break
        x = x + delta_x
        if np.linalg.norm(delta_x) < tol:
            print(f"Converged in {i+1} iterations.")
            return x
    print("Maximum iterations reached without convergence.")
    return x
```

```python
def main(): 
    f = input_function()
    x = input_vector()
    print(newton_raphson(f,x))
```

```python
main()
```
